```python
import pandas as pd
d = "data/fatmawati_v3/"

# 1. What does each column mean? Read this first.
dic = pd.read_csv(d + "data_dictionary.csv")
dic[dic.table == "episodes"]
```

```python
# 2. One row per simulated hour: the easiest table to start with
ep = pd.read_csv(d + "episodes.csv")
ep.head()
```

```python
# 3. Compare controllers: average delay per demand level (a pivot table)
ep.pivot_table(index="controller", columns="level",
               values="stopped_delay_s", aggfunc="mean").round(1)
```

```python
# 4. Ambulance trips, joined to their episode's controller and level
amb = pd.read_csv(d + "ambulance_trips.csv")
amb = amb.merge(ep[["episode_id", "controller", "level"]], on="episode_id")
amb.groupby(["level", "controller"])["travel_time_s"].agg(["count", "mean", "median"]).round(1)
```

```python
# 5. Vehicle mix: count every vehicle entering the junction once
loops = pd.read_csv(d + "loops_1min.csv")
mix = loops[loops.kind == "entry_gate"][["n_motorcycle", "n_car", "n_bus", "n_truck"]].sum()
(100 * mix / mix.sum()).round(1)          # should be about 80% motorcycles
```
